# gradient-descent-animation(both-m-and-b)

```python

from sklearn.datasets import make_regression

import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from matplotlib.animation import FuncAnimation
import matplotlib.animation as animation
from IPython.display import HTML
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
b = -520
m = 600
lr = 0.001
all_b = []
all_m = []
all_cost = []

epochs = 30

for i in range(epochs):
    slope_b = 0
    slope_m = 0
    cost = 0
    for j in range(X.shape[0]):
        slope_b = slope_b - 2*(y[j] - (m * X[j]) -  b)
        slope_m = slope_m - 2*(y[j] - (m * X[j]) -  b)*X[j]
        cost = cost + (y[j] - m * X[j] -b) ** 2

    b = b - (lr * slope_b)
    m = m - (lr * slope_m)
    all_b.append(b)
    all_m.append(m)
    all_cost.append(cost)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML


fig, ax = plt.subplots(figsize=(9,5))

# data points
ax.scatter(X, y)

# x range for line
x_i = np.arange(-3, 3, 0.1)

# initial line
line, = ax.plot(
    x_i,
    x_i * all_m[0] + all_b[0],
    'r-',
    linewidth=2
)


def update(i):

    # update regression line
    y_i = x_i * all_m[i] + all_b[i]
    

    line.set_ydata(y_i)

    ax.set_title(f"Epoch : {i+1}")

    return line,


# create animation
anim = FuncAnimation(
    fig,
    update,
    frames=len(all_m),
    interval=500,
    repeat=False
)


# display animation in notebook
HTML(anim.to_jshtml())
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML


# epochs
num_epochs = list(range(len(all_cost)))


fig, ax = plt.subplots(figsize=(9,5))

# axis limits
ax.set_xlim(0, len(all_cost))
ax.set_ylim(0, max(all_cost) + 100000)

# empty line
line, = ax.plot([], [], linewidth=2)

xdata = []
ydata = []


def animate(i):

    xdata.append(num_epochs[i])
    ydata.append(all_cost[i])

    line.set_data(xdata, ydata)

    ax.set_xlabel(f"Epoch : {i+1}")
    ax.set_ylabel("Cost")

    return line,


# create animation
anim = FuncAnimation(
    fig,
    animate,
    frames=len(all_cost),
    interval=500,
    repeat=False
)


# show animation in notebook
HTML(anim.to_jshtml())
```
