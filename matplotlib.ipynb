# matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x=[2,3,4,5,7]
y=[23,58,90,56,32]
plt.plot(x,y)
plt.xlabel("values",color='red')
plt.ylabel("random values")

plt.show()
```

```python
student=["A","B","c"]
marks=[45,89,90]
plt.bar(student,marks)
plt.xlabel("student")
plt.ylabel("marks")

plt.show()
```

```python
x=[2,3,4,5,7]
y=[23,58,90,56,32]
plt.scatter(x,y)
plt.xlabel("values",color='red')
plt.ylabel("random values")

plt.show()
```

```python
x=[2,3,4,5,7]
y=[23,58,90,56,32]
plt.hist(x,y,bins=2)
plt.xlabel("values",color='red')
plt.ylabel("random values")
# tells about distribution of data
plt.show()
```
