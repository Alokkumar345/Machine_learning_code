# working with dates

```python
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

# extract year

```python
date['date_year']=date['date'].dt.year
```

```python
# date.head()
date.sample(5)
```

# extract month

```python
date['date_month_no']=date['date'].dt.month
```

```python
date.head(3)
```

```python
date['date_month_no']=date['date'].dt.month_name()
```

```python
date.head()
```

# extract days

```python
date['date_day']=date['date'].dt.day
```

```python
date.head()
```

```python
# day of week
date['date_day_of_week']=date['date'].dt.dayofweek
```

```python
date.head()
```

```python
# day of week name
date['day_dow_name']=date['date'].dt.day_name()
```

```python
date.head()
```

```python


# is weekend?

date['date_is_weekend'] = np.where(date['day_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

# extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id','city_id','orders']).head()
```

# extract quarter

```python
date['quarter']= date['date'].dt.quarter
date.drop(columns=['product_id','city_id','orders']).head()
```

# extract semester

```python
date['semester']=np.where(date['quarter'].isin([1,2]),1,2)
date.drop(columns=['product_id','city_id','orders']).head()
```

# extract time elapsed b/w date

```python
import datetime
today=datetime.datetime.today()
```

```python
today
```

```python
today-date['date']
```

```python
(today-date['date']).dt.days
```

```python
# months passed

np.round((today - date['date']).dt.days / 30, 0)
```

# time

```python
time.info()
```

```python
time['date']=pd.to_datetime(time['date'])
```

```python

time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

# extract time part

```python

time['time'] = time['date'].dt.time

time.head()
```

# time diff

```python

today - time['date']
```

```python

# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python

# in hours

(today - time['date'])/np.timedelta64(1,'h')
```
